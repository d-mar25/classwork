```python
from datetime import datetime
this_year = datetime.now().year
driving_age = 16
is_year_finished = False

name = input("What is your name? ______")

while not is_year_finished:
    birth_year = input("What is your birth year? _____")
    if not birth_year.isnumeric():
        print("birth year is not numeric. Must use numbers")
        continue
    if len(birth_year) != 4:
        print("birth year must be 4 digits long")
        continue
    if (not birth_year.startswith("19")) and (not birth_year.startswith("20")):
        print(birth_year.startwith("19"), birth_year.startswith("20"))
        print(f"{birth_year} Must be born in the 20th or 21st century")
        continue

    birth_year = int(birth_year)
    is_year_finished = True

person_age = this_year - birth_year

person_age

if person_age > driving_age:
    print("you can drive")
else:
    print("you cannot drive")
```

```python
def calculate_age(birth_year: int) -> int:
    """
    This function calculates the age based on the birth year given
    
    Arg:
        birth_year(int): 
    """
    this_year = datetime.now().year
    age = this_year - birth_year
    return age

age = calculate_age(2007)
age
```
