# Data types

## Simple Data types
    - Simplest ways to store information

- string (str)
- integer (int)
- float (float)
    - decimalized number 
        - i.e. 3.0
    - Binary string for integers are different from binary for floats
- boolean (bool)
    - T/F

## Complex Data types
    - To store a collection information expressed by simple data types
    - "Building blocks of programming"

- list (list)
- typle (tuple)
- dictionary (dict)
- set (set)

## Dates

- Dates and datetime

```python
# Integers
x: int = 26
y: int = 5

print("addition: ", x + y) # addition
print(f"division: {x/y}") # division
print(f"multiplication: {x*y}") # multiplication
print(f"subtraction: {x-y}") # subtraction
print(f"exponentiation: {x**y}") # exponentiation
print(f"floor division: {x//y}") # floor division
print(f"modulus: {x % y}") # modulus
```

```python
x = str(x)
y = float(y)

print(f" new type x and y: {x.zfill(3), y}")
print(int(3.99999))
```

```python
# booleans

a: bool = True
print(f"x = {x}")
print(f"y as bool: {bool(y)}")
print(f"0 as bool: {bool(0)}")
print(f"empty string as bool: {bool('')}")
print(f"x string as bool: {bool(x)}")
```
