```python
#Chess with lists

pawns = ["P"] * 8
cool = ["R", "N", "B", "Q", "K", "B", "N", "R"]
empty = [" "] * 8

chess = [
    cool.copy(),
    pawns.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    pawns.copy(),
    cool.copy(),
]
chess
```

```python
chess[0][1] = " "
chess[2][2] = "N"
chess
```

```python
# tuple

# not mutable (CANNOT CHANGE)
# allows duplicate values
# indexed 
tpl = (1,2,2,3,1,"something")
tpl2 = tuple((1,2,3,4,5))
print(tpl)
print(tpl2)
```

```python
tpl.count(1)
```

```python
# set
# can change (mutable)
# no duplicates
    # displays only unique values
# unordered

set1 = {1,2,1,2,1,2,3} # bad way
set2 = set([1,2,3,4,5,4,4,3,3]) # good way
set2
```

```python
len(set2)
```

```python
set2.add(6)
len(set2)
```

```python
set2.remove(3)
set2.remove(4)
len(set2)
```

```python
set2
```

```python
set2.difference(set1)
set2 - set1
```

```python
# dictionary 
# key: value pairs
# indexed by its key (not by its position)
# can change (mutable)
# keys cannot duplicate, values can.

person = {
    "name": "Dom",
    "age": 19,
    "city": "Burnaby",
    "family": [
        "mom", 
        "dad", 
        "brother",
    ],
}
person
```

```python
person["city"]
```

```python
person["family"][2]
```

```python
name = input("Enter name here...")
age = input("Enter age here...")
city = input("Where do you live?")

person = {
    "name": name,
    "age": age,
    "city": city,
}

print(f"{person['name']} is {person['age']} years old and lives in {person['city'].title()}")
```
