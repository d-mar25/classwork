```python
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
from scipy import stats
# import scipy.stats
url = "https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/soccer-players.csv"
# https://raw.githubusercontent.com/nkmwicz/data-for-students/
# refs/heads/main/soccer-players.csv

df = pd.read_csv(url)
df.head()
```

```python
plt.figure(figsize=(3,2))
plt.xlabel("Day")
plt.ylabel("Sales ($)")
plt.title("Fig 1: Daily Coffee Sales")
plt.xlim(0,7)
plt.ylim(0,60)
plt.grid(True)
txt = "Sales Item"
#plt.text(1, 20, txt, ha ="center", va="center")
plt.text(
    0.20,
    0.70,
    txt,
    ha="center",
    color="red",
    transform=plt.gca().transAxes #transforms raw values into percentage values of the axes
    )
plt.show()
```

```python
plt.figure(figsize=(3,2))
bins = df["height_in_cm"].count() ** (1/3)
plt.hist(df['height_in_cm'], bins=10, edgecolor='black')
plt.show()
```

```python
plt.figure(figsize=(8,6))
hist = sns.histplot(data=df, x="height_in_cm", kde = True)
plt.ylabel("Frequency")
hist.set(xlabel="Height (cm)")
mean_val = df['height_in_cm'].mean()
median_val = df['height_in_cm'].median()
plt.axvline(
    mean_val,
    color='red',
    linestyle='dashed',
    linewidth=1,
    label='Mean'
    )
plt.axvline(
    median_val,
    color='yellow',
    linestyle='--',
    label='Median'
)

skew = stats.skew(df['height_in_cm'], nan_policy='omit')
kurt = stats.kurtosis(df['height_in_cm'], nan_policy='omit')
txt = f"""Skew: {skew: .2f}
Kurt: {kurt: .2f}
"""
plt.text(
    0.05,
    0.60,
    txt,
    transform=plt.gca().transAxes,
)
hist.set(title='Histogram Soccer Player Heights')
sns.set_theme(style="whitegrid")
plt.legend()
plt.show()
```

```python
# Plt boxplot
plt.figure(figsize=(8,6))
plt.boxplot(x=df['height_in_cm'].dropna(), orientation='horizontal')
plt.show()
```
