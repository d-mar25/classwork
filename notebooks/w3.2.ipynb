# Controlling Python Flow

```python
from datetime import datetime
from typing import TypedDict
# comparisons
# ==, != . < , > , <= , >=
class Person(TypedDict):
    name: str
    birth_year: int
    
drinking_age: int = 21
current_year: int = datetime.now().year

name: str = input("Enter your name..."),
birth_year: str = input("Enter your birth year...")


person1: Person = {
    "name": name,
    "birth_year": birth_year,
}
if birth_year.isnumeric():
    person1['birth_year'] = int(birth_year)
    
    print(f"Is {person1['name']} old enough to drink? {(current_year - person1['birth_year']) >= drinking_age}")
else:
    print("A year is an integer, dummy!")
```

```python
if person1['birth_year'] and type(person1['birth_year']) == str:
    print("birth_year exists and is a string")
elif person1['birth_year'] and type(person1['birth_year']) == int:
    print("birth_year exists and birth_year is an int")
else:
    print("birth_year is either not existing or not a string")
```
