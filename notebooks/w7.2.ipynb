```python
import pandas as pd
from scipy import stats
# import scipy.stats
url = "https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/soccer-players.csv"
# https://raw.githubusercontent.com/nkmwicz/data-for-students/
# refs/heads/main/soccer-players.csv

df = pd.read_csv(url)
df.head()
```

```python
df['height_in_cm'].std()
```

```python
df['height_in_cm'].skew()
```

```python
df['height_in_cm'].min()
```

```python
# 1cm = 0.39in
inch_ratio = 0.39

df['height_in_ft'] = (df['height_in_cm'] * inch_ratio) / 12
df['height_in_ft'].mean()
```

```python
df.info()
```

```python
for col in df.columns:
    print(col)
```

```python
skew = stats.skew(df["height_in_cm"], nan_policy="omit")
kurt = stats.kurtosis(df["height_in_cm"], nan_policy="omit")
print(f"skew = {skew}")
print(f"Kurtosis = {kurt}")
```

```python
df['date_of_birth'] = pd.to_datetime(df["date_of_birth"])
```

```python
df['birth_mth'] = df['date_of_birth'].dt.month
df['birth_mth'].skew()
```

```python
df["birth_mth"].kurt()
```

```python
for x in range(1,13):
    x = float(x)
    perc = df['birth_mth'].value_counts().loc[x] / df['birth_mth'].count()
    print(x, perc)
```
